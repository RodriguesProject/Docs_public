# Converter texto em data

Datas podem como texto. Fazemos isso para calcular prazos.



```python
from datetime import datetime

texto = "01/09/2026"

data = datetime.strptime(texto, "%d/%m/%Y")

print(data)
print(data.year)
```

# Criar uma cópia e converter as colunas

```python
df = df_csv.copy()
```

```python
df["data_pedido"] = pd.to_datetime(
    df["data_pedido"],
    format="%Y-%m-%d",
    errors="coerce"
)

df["data_entrega"] = pd.to_datetime(
    df["data_entrega"],
    format="%Y-%m-%d",
    errors="coerce"
)
```

```python
print(df[["data_pedido", "data_entrega"]].dtypes)
print(df[["data_pedido", "data_entrega"]].isna().sum())
```

# Criar colunas de ano, mês e dia da semana

```python
df["ano"] = df["data_pedido"].dt.year
df["mes"] = df["data_pedido"].dt.month
df["dia_semana"] = df["data_pedido"].dt.day_name()
```

```python
display(
    df[["pedido_id", "data_pedido", "ano", "mes", "dia_semana"]]

)
```

# Calcular o prazo de entrega

"Um pedido feito em 1 de setembro e entregue e, 5 de setembro, ou seja, ele levou quatro dias ocorridos."

```python
df["prazo_entrega"] = (
    df["data_entrega"] - df["data_pedido"]
).dt.days
```

```python
display(
    df[
        ["pedido_id",
         "status",
         "data_pedido",
         "prazo_entrega"
         ]
        ]
)
```
