# Aritméticos

"+" = somar
"-" = subtração

"*" = multiplicação

/ = divisão - resultado decimal

% = usa para saber se o número é par ou ímpar; qual é o resto da divisão

// = divisão inteira - números inteiros

"**" = elevar um número a outro

```python
valor = 200
desconto = 20

valor_final = valor - desconto

print(f"Valor final: R$ {valor_final}")
```

# Relacionais

# Lógicos

* and -"E se" - as duas condições precisam ser verdadeiras

* or - signica "ou" - basta uma condição ser verdadeira

```python
idade = 18
tem_carteira = True

if idade >= 18 and tem_carteira == True:
    print("Pode dirigir")
```
