```python
# verificar os espaços

colunas_texto = [
    'CL_GENERO',
    'CL_SEG',
    'PR_CAT',
    'PR_NOME',
    ]

for coluna in colunas_texto:
    quantidade_em_branco = (
        df[coluna]
        .astype("string")
        .str.strip()
        .eq("")
        .sum()
    )

print(coluna, ":", quantidade_em_branco)
```

```python
colunas_alvo = ['CL_GENERO', 'CL_SEG', 'PR_CAT', 'PR_NOME']

# Verifica para cada linha se alguma das colunas selecionadas está vazia ou contém apenas espaços
clientes_em_branco = (
    df[colunas_alvo]
    .astype("string")
    .apply(lambda x: x.str.strip() == "")
    .any(axis=1)
)

print(
    "Clientes preenchidos apenas com espaços em alguma das colunas: ",
    clientes_em_branco.sum()
)

display(
    df.loc[
        clientes_em_branco,
        colunas_alvo
    ]
)
```
