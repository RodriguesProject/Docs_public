# Aula Numpay - Operações em Numpy

Numpy é uma biblioteca de cálculo numérico com arrays.

* Array é uma estrutura que guarda vários valores em sequência, organizados sob um mesmo nome.

Exemplo:

1D = vetor = uma linha de valores, como uma lista
2D = matriz = linhas e colunas, como uma planilha
3D ou + = camadas de matrizes(exemplo: largura x altura x cor)

v = np.array([1, 2, 3]) # 1D

m = np.array([1, 2, 3]), [4, 5, 6]) 2D 2 linhas, 3 colunas

m[1, 2] # 6 (linha 1, coluna 2)

m.shape #(2, 3)
