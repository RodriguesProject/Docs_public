# Transformando uma política de desconto em decisões.

- Quando o valor da compra for a partir de R$ 200
desconto = 10%

- Quando o valor da compra for a partir de R$ 100

e abaixo de R$ 200
desconto = 5%

- abaixo de R$ 100
não tem desconto.

## Uma compra de R$ 235 deve receber quanto de desconto?

```python
valor = 235.0

# Verifique primeiro a faixa de maior desconto
if valor >= 200:
  desconto = 0.10

# Esta condição só é avaliada se o primeiro teste for falso.
elif valor >= 100:
  desconto = 0.05

# Atende aos valores que não entraram nas faixas anteriores.
else:
  desconto = 0

# Calcular a parte do valor que o cliente deve pagar
valor_final = valor * (1 - desconto)
print(f"Valor final: R$ {valor_final: .2f}")
```

# Repetição com While

Simulação de tentativas de senha. O programa de parar quando encontrar "dados123" e quando chegar a 3 tentativas.

```python
tentativas = ["123", "senha", "dados123"]
indice = 0
autenticado = False

while indice < len(tentativas) and indice < 3 and not autenticado:
  if tentativas[indice] == "dados123":
    autenticado = True

  else:
    indice += 1

print("Acesso liberado!" if autenticado else "Acesso bloqueado")
```

"123" = 0 = que a comparação é falsa; o indíce passa para 1.

"senha" = a comparação é falsa; então passe para o índice 2

"dados123" = autenticado passa a ser True; termina aqui o laço de repetição

# Listas, Tuplas, Dicionários e estruturas compostas

**objetivo**: escolher a estrutura mais adequada para organizar os dados antes de usar as biblitecas como o Pandas.

# 1. Lista

Listas são mutáveis, e servem para organizar uma sequência.

Permite adicionar, remover e substituir elementos.

os índices dos valores na lista sempre começam em zero.

```python
vendas = [120.0, 85.5, 230.0, 99.9]

# Consulta o primeiro elemento
print(vendas[0])

# Adiciona uma venda ao final da lista
vendas.append(320.10)

# Conta os elementos da lista
print("Quantidade: ", len(vendas))

# Soma valores e apresenta casas decimais
print(f"Soma: R$ {sum(vendas): 2f}")
```

# 2. Tupla

Serve para informações que representam um conjunto fixo.

```python
coordenada = (-22.90, -43.20)

# Distribui os dois elementos em duas variáveis
latitude, longitude = coordenada

print("Latitude: ", latitude)
print("Longitude: ", longitude)
```

# Dicionário

dicionários servem para organizar informações por chave.

```python
produto = {
    "nome": "Teclado",
    "preco": 249.90,
    "estoque": 7
}

print(produto["preco"])

# Atualiza um campo existente
produto["estoque"] = 6

# Adiciona um novo campo
produto["categoria"] = "Periféricos"

print(produto)
```

## Conclusão:

lista reune os produtos. Dicionário descreve um produto. Em cada repetição, "item" recebe um desses dicionários completos.
