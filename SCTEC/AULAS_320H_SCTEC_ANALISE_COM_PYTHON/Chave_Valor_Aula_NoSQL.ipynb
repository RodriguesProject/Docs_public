| Informação | Proposta para o caso | Motivo |
|---|---|---|
| Itens do pedido | Embutir no pedido | São consultados juntos e pertencem à compra |
| Preço na compra | Preservar no item | Mudanças de preço no catálogo não devem alterar o histórico |
| Cadastro atual do cliente | Pode ficar separado | Tem atualizações próprias |
| Nome e UF registrados na compra | Podem ser cópias históricas | Preservam o contexto do pedido |
| Histórico que cresce indefinidamente | Considerar coleção separada | O crescimento exige outra organização |
