# Branch

É uma linha paralela de trabalho, na qual é usada pra trabalhar sem quebrar a versão principal

```python
!git checkout -b feature/calculadora
```

```python
%%writefile calculadora.py
a = 10
b = 4

print("soma:", a + b)
print("subtração:", a - b)
```

```python
!git add calculadora.py
!git commit -m "Adiciona calculadora"
```

# Merge

Branch = linha paralela de trabalho

Merge = integração das mudanças

```python
!git checkout main
!git merge feature/calculadora
```

```python
!git log --oneline --graph --decorate --all
```

# Resumo do mini-case

1. **Branches** Isolam trabalhos paralelos para reduzir risco de alterar a versão principal antes da hora.

2. **Commit** é um registro versionado de um conjunto coerente de mudanças.

3. **Git** é o sistema de controle de versões do nosso projeto; **Github** hospeda os nossos repositórios e adiciona colaboração, como por exemplo Pull Requests, Issues, etc.
